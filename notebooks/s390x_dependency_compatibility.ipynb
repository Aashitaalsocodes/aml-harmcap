```python
from Crypto.PublicKey import ECC
from Crypto.Signature import eddsa
key = ECC.generate(curve="Ed25519")
sig = eddsa.new(key, "rfc8032").sign(b"hello")
eddsa.new(key.public_key(), "rfc8032").verify(b"hello", sig)
print("ed25519 ok")
```
